Practicing the GrAPE Algorithm
==============================

```julia
using GRAPE
using QuantumPropagators
```

Starting with our normal 2-level system we can define the Hamiltonian:
$$
\hat{H} = 
\left[
    \begin{matrix}
    0 & \Omega/2 \\
    \Omega/2 & -\Delta \\
    \end{matrix}
\right]
$$


```julia
using QuantumPropagators
using StaticArrays

basis = [
    SVector{2}(ComplexF64[1, 0]),  # |0⟩
    SVector{2}(ComplexF64[0, 1]),  # |1⟩
];

target_states = [
    SVector{2}(ComplexF64[0, 1]),  # |0⟩ -> |1⟩
    SVector{2}(ComplexF64[1, 0]),  # |1⟩ -> |0⟩
]

matrixΩ = SMatrix{2,2,ComplexF64}(
    [0       1/2
     1/2     0]
)

matrixΔ = SMatrix{2,2,ComplexF64}(
    [0      0
     0     -1]
)

H(Δ, Ω) = hamiltonian(0, (matrixΔ, Δ), (matrixΩ, Ω))
```

```julia
using QuantumPropagators: propagate, Cheby

pops = propagate(
    basis[2], H(1, 1), tlist; method=Cheby, storage=true, observables=(Ψ -> Array(abs2.(Ψ)), )
)

fig = plot(tlist ./ ns, pops'; label=["0" "1"], xlabel="time (ns)", legend=:outertop, legend_column=-1)
```
